### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)


tienda.head()
```

#1. Análisis de facturación



```python
import numpy as np
import matplotlib.pyplot as plt

precios=[tienda['Precio'],tienda2['Precio'],tienda3['Precio'],tienda3['Precio']]
nombre_tiendas=["tienda 1","tienda 2","tienda 3","tienda 4"]
i=0
ingresos_totales=[]

for i,pre in enumerate(precios):
  print(f"Ingreso total de la {nombre_tiendas[i]}")
  print(pre.sum())
  ingresos_totales.append(pre.sum())
  print()
  print()

# ingreso total combinado
total_ingresos = np.sum(ingresos_totales)

#  proporción de cada tienda
proporciones = [ingreso / total_ingresos for ingreso in ingresos_totales]

# gráfico
plt.figure(figsize=(8, 8))
plt.pie(proporciones, labels=nombre_tiendas, autopct='%1.1f%%', startangle=140)
plt.title('Proporción de Ingresos Totales por Tienda')
plt.axis('equal')
plt.show()
```

Observamos que los ingresos de la tienda varían entre \$1,098,019,600 (24.6 %) a \$1,150,880,400 (25.8%) en donde se muestra que la tienda 1 tiene un mejor rendimiento a las demás.

# 2. Ventas por categoría

### Ventas por categoría de cada tienda

```python
import matplotlib.pyplot as plt

categorias=[tienda['Categoría del Producto'],tienda2['Categoría del Producto'],tienda3['Categoría del Producto'],tienda4['Categoría del Producto']]
nombre_tiendas=["tienda 1","tienda 2","tienda 3","tienda 4"]

for i, cat in enumerate(categorias):
  print(nombre_tiendas[i])
  conteo_por_categoria = cat.value_counts()
  print(f"Electrónicos:{conteo_por_categoria.get('Electrónicos', 0)}")
  print(f"Muebles:{conteo_por_categoria.get('Muebles', 0)}")
  print(f"Electrodomésticos:{conteo_por_categoria.get('Electrodomésticos', 0)}")
  print(f"Juguetes: {conteo_por_categoria.get('Juguetes', 0)}")
  print(f"Artículos para el hogar: {conteo_por_categoria.get('Artículos para el hogar', 0)}")
  print(f"Libros: {conteo_por_categoria.get('Libros', 0)}")
  print(f"Deportes y diversión: {conteo_por_categoria.get('Deportes y diversión', 0)}")
  print(f"Instrumentos musicales: {conteo_por_categoria.get('Instrumentos musicales', 0)}")
  #gráfico
  plt.figure(figsize=(10, 6))
  conteo_por_categoria.plot(kind='bar',color='red')
  plt.title(f"Ventas por categoría en {nombre_tiendas[i]}")
  plt.xlabel("Categoría")
  plt.ylabel("Cantidad Vendida")
  plt.xticks(rotation=45, ha='right')
  plt.tight_layout()
  plt.show()
  print()
```

# 4. Productos más y menos vendidos

```python
import matplotlib.pyplot as plt

productos=[tienda['Producto'],tienda2['Producto'],tienda3['Producto'],tienda4['Producto']]
nombre_tiendas=["tienda 1","tienda 2","tienda 3","tienda 4"]

for i,p in enumerate(productos):
  print(nombre_tiendas[i])
  conteo_productos=p.value_counts()
  print()
  print("Productos más vendidos")
  print(conteo_productos.head())
  print()
  print("Productos menos vendidos")
  print(conteo_productos.tail())
  print()
  print()

```

# 5. Envío promedio por tienda

```python
import numpy as np
tiendas=[tienda,tienda2,tienda3,tienda4]
i=0
costos_totales=[]
for i,t in enumerate(tiendas):
  conteo_de_envios=t['Costo de envío']
  print(f"Costo de envío de tienda {i+1} : {round(conteo_de_envios.mean(),2)}")
  costos_totales.append(round(conteo_de_envios.mean(),2))
  print()
# ingreso total combinado
total_costos = np.sum(costos_totales)

#  proporción de cada tienda
proporciones = [costo / total_costos for costo in costos_totales]

# gráfico
plt.figure(figsize=(8, 8))
plt.pie(proporciones, labels=nombre_tiendas, autopct='%1.1f%%', startangle=140)
plt.title('Proporción de Ingresos Totales por Tienda')
plt.axis('equal')
plt.show()
```

Se observa que el costo promedio de envío en las 4 tiendas de la cadena de Alura Store varía entre \$23,460 (23.6%) a \$26,018(26.1%).
En este caso, se observa que la tienda 4 tiene un mejor costo promedio de envío y la tienda 1 tiene el costo promedio de envío más alto de las 4 tiendas.
